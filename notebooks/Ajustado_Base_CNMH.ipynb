```python
file_path = '/content/drive/MyDrive/Proyecto de profundización 1/PROYECTO GRADO 2025/Resultados modelos/CasosMA_202409.xlsx'

datamasacres = pd.read_excel(file_path, na_values="-", index_col=5, engine='openpyxl')
datamasacres
```

```python
datamasacres.Departamento=datamasacres.Departamento.str.title()
datamasacres.Municipio=datamasacres.Municipio.str.title()
datamasacres.Región=datamasacres.Región.str.title()
datamasacres.Modalidad=datamasacres.Modalidad.str.title()
datamasacres["Presunto Responsable"] = datamasacres["Presunto Responsable"].str.title()
datamasacres.head()
```

```python
datamasacres.info()
```

```python
datamasacres = datamasacres[(datamasacres["Año"] >= 1990) & (datamasacres["Año"] <= 2023)]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Cantidad de víctimas por departamento
victimas_depto = datamasacres.groupby("Departamento")["Total de Víctimas del Caso"].sum().reset_index()

plt.figure(figsize=(12,6))
ax = sns.barplot(
    data=victimas_depto.sort_values("Total de Víctimas del Caso", ascending=False),
    x="Departamento", y="Total de Víctimas del Caso", palette="Reds_r"
)

# Etiquetas dentro de cada barra
for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", label_type="edge", fontsize=9, color="black")

plt.xticks(rotation=90)
plt.title("Total de víctimas por departamento")
plt.ylabel("Número de víctimas")
plt.xlabel("Departamento")
plt.tight_layout()
plt.show()


# Cantidad de registros por departamento
registros_depto = datamasacres["Departamento"].value_counts().reset_index()
registros_depto.columns = ["Departamento", "Cantidad de Registros"]

plt.figure(figsize=(12,6))
ax = sns.barplot(
    data=registros_depto,
    x="Departamento", y="Cantidad de Registros", palette="Blues_r"
)

# Añadir etiquetas en cada barra
for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", label_type="edge", fontsize=9, color="black")
plt.xticks(rotation=90)
plt.title("Cantidad de registros por departamento")
plt.ylabel("Número de registros")
plt.xlabel("Departamento")
plt.tight_layout()
plt.show()
```

```python
col_mod = next(c for c in datamasacres.columns
               if c.strip().casefold() == "modalidad".casefold())

posibles_vict = {
    "total de víctimas del caso", "total de victimas del caso",
    "víctimas", "victimas", "n_victimas", "num_victimas"
}
col_vic = next(c for c in datamasacres.columns
               if c.strip().casefold() in {s.casefold() for s in posibles_vict})

# Limpiar y sumar
df = datamasacres[[col_mod, col_vic]].copy()
df[col_mod] = df[col_mod].fillna("Sin información").astype(str).str.strip()
df[col_vic] = pd.to_numeric(df[col_vic], errors="coerce").fillna(0)

victimas_por_mod = (
    df.groupby(col_mod, as_index=False)[col_vic]
      .sum()
      .sort_values(col_vic, ascending=False)
)

# Renombrar para el gráfico
victimas_por_mod.columns = ["Modalidad", "Total de Víctimas del Caso"]

# Graficar (horizontal)
plt.figure(figsize=(10, 8))
ax = sns.barplot(data=victimas_por_mod, x="Total de Víctimas del Caso", y="Modalidad", palette="Blues_r")
for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", label_type="edge", fontsize=9)
plt.title("Total de Víctimas del Caso por Modalidad")
plt.xlabel("Total de Víctimas del Caso")
plt.ylabel("Modalidad")
plt.tight_layout()
plt.show()


col_mod = next(c for c in datamasacres.columns
               if c.strip().casefold() == "modalidad".casefold())

posibles_vict = {
    "total de víctimas del caso", "total de victimas del caso",
    "víctimas", "victimas", "n_victimas", "num_victimas"
}
col_vic = next(c for c in datamasacres.columns
               if c.strip().casefold() in {s.casefold() for s in posibles_vict})

# Limpiar y sumar
df = datamasacres[[col_mod, col_vic]].copy()
df[col_mod] = df[col_mod].fillna("Sin información").astype(str).str.strip()
df[col_vic] = pd.to_numeric(df[col_vic], errors="coerce").fillna(0)

victimas_por_mod = (
    df.groupby(col_mod, as_index=False)[col_vic]
      .sum()
      .sort_values(col_vic, ascending=False)
)

# Renombrar para el gráfico
victimas_por_mod.columns = ["Modalidad", "Total de Víctimas del Caso"]

# Graficar (horizontal)
plt.figure(figsize=(10, 8))
ax = sns.barplot(data=victimas_por_mod, x="Total de Víctimas del Caso", y="Modalidad", palette="Blues_r")
for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", label_type="edge", fontsize=9)
plt.title("Total de Víctimas del Caso por Modalidad")
plt.xlabel("Total de Víctimas del Caso")
plt.ylabel("Modalidad")
plt.tight_layout()
plt.show()
```

```python
# Group by year and polarity, and count the number of songs
annual_polarity_summary = Polaridad.groupby(['AÑO', 'POLARIDAD']).size().unstack(fill_value=0)

print("Serie anual de canciones por polaridad:")
display(annual_polarity_summary)
```

```python
# tabla combinada
if 'datamasacres' not in locals():
    print("datamasacres DataFrame not found. Please ensure the cell loading datamasacres is run.")
    annual_summary = pd.DataFrame(columns=['Año', 'Total_Casos', 'Total_Victimas'])
else:
    if 'Año' in datamasacres.columns:
        datamasacres['Año'] = datamasacres['Año'].astype(int)
    else:
        datamasacres.index = datamasacres.index.astype(int)
        datamasacres = datamasacres.reset_index()

    annual_summary = datamasacres.groupby('Año').agg(
        Total_Casos=('ID Caso', 'count'),
        Total_Victimas=('Total de Víctimas del Caso', 'sum')
    ).reset_index()

if 'annual_polarity_summary' in locals():
    annual_polarity_summary_reset = annual_polarity_summary.reset_index()
    annual_polarity_summary_reset = annual_polarity_summary_reset.rename(columns={'AÑO': 'Año'})
else:
    print("annual_polarity_summary DataFrame not found. Please ensure the cell generating annual_polarity_summary is run.")
    annual_polarity_summary_reset = pd.DataFrame()

if not annual_polarity_summary_reset.empty and not annual_summary.empty:
    combined_data = pd.merge(annual_polarity_summary_reset, annual_summary, on='Año', how='left')

    unified_table = combined_data[['Año', 'Negativo', 'Neutro', 'Positivo', 'Total_Casos', 'Total_Victimas']]

    print("Tabla unificada de año, polaridad de canciones y masacres por año:")
    display(unified_table)
else:
    print("Could not generate the combined table because one or both of the required DataFrames are missing.")
```

```python
import matplotlib.pyplot as plt

unified_table['Año'] = unified_table['Año'].astype(int)
unified_table = unified_table.sort_values('Año')

# Crear figura y primer eje (para barras apiladas)
fig, ax1 = plt.subplots(figsize=(14, 6))

# Posiciones para el eje X
x = unified_table['Año']

# Barras apiladas
ax1.bar(x, unified_table['Negativo'], label='Negativo', color='salmon')
ax1.bar(x, unified_table['Neutro'], bottom=unified_table['Negativo'], label='Neutro', color='tan')
ax1.bar(x, unified_table['Positivo'],
        bottom=unified_table['Negativo'] + unified_table['Neutro'],
        label='Positivo', color='darkseagreen')

# Eje izquierdo
ax1.set_ylabel('Número de canciones por polaridad')
ax1.set_xlabel('Año')
ax1.set_xticks(x)
ax1.set_xticklabels(x, rotation=90)
ax1.legend(loc='upper left')
ax1.grid(axis='y', linestyle='--', alpha=0.5)

# Eje derecho: Total de casos
ax2 = ax1.twinx()
ax2.plot(x, unified_table['Total_Victimas'], color='blue', marker='o', linestyle='-', label='Total de víctimas')
ax2.set_ylabel('Total de víctimas')
ax2.legend(loc='upper right')

# Título general
plt.title('Total de víctimas y polaridad de las producciones musicales por año (1990–2023)')
plt.tight_layout()
plt.show()
```

```python
# Group by year and polarity, and count the number of songs
annual_emocion_summary = Emociones.groupby(['AÑO', 'EMOCION']).size().unstack(fill_value=0)

print("Serie anual de canciones por emociones:")
display(annual_emocion_summary)
```

```python
# tabla combinada
if 'datamasacres' not in locals():
    print("datamasacres DataFrame not found. Please ensure the cell loading datamasacres is run.")
    annual_summary = pd.DataFrame(columns=['Año', 'Total_Casos', 'Total_Victimas']) # Create an empty DataFrame to prevent errors
else:
    if 'Año' in datamasacres.columns:
        datamasacres['Año'] = datamasacres['Año'].astype(int)
    else:
        datamasacres.index = datamasacres.index.astype(int)
        datamasacres = datamasacres.reset_index() # Make 'Año' a regular column for easier grouping

    annual_summary = datamasacres.groupby('Año').agg(
        Total_Casos=('ID Caso', 'count'),
        Total_Victimas=('Total de Víctimas del Caso', 'sum')
    ).reset_index()

if 'annual_emocion_summary' in locals():
    annual_emocion_summary_reset = annual_emocion_summary.reset_index()
    annual_emocion_summary_reset = annual_emocion_summary_reset.rename(columns={'AÑO': 'Año'})
else:
    print("annual_emocion_summary DataFrame not found. Please ensure the cell generating annual_emocion_summary is run.")
    annual_emocion_summary_reset = pd.DataFrame() # Create an empty DataFrame to prevent errors

if not annual_emocion_summary_reset.empty and not annual_summary.empty:
    combined_data = pd.merge(annual_emocion_summary_reset, annual_summary, on='Año', how='left')

    unified_table = combined_data[['Año', 'Tristeza', 'Alegría', 'Ira', 'Miedo','Asco','Sorpresa','Total_Casos', 'Total_Victimas']]

    print("Tabla unificada de año, emociones de canciones y masacres por año:")
    display(unified_table)
else:
    print("Could not generate the combined table because one or both of the required DataFrames are missing.")
```

```python
import matplotlib.pyplot as plt

unified_table['Año'] = unified_table['Año'].astype(int)
unified_table = unified_table.sort_values('Año')

fig, ax1 = plt.subplots(figsize=(18, 6))
x = unified_table['Año']

emociones = ["Alegría", "Tristeza", "Ira", "Miedo", "Asco", "Sorpresa"]
colores   = ["salmon", "skyblue", "darkkhaki", "darkseagreen", "olive", "lightgreen"]

bottom = None
for emo, col in zip(emociones, colores):
    if bottom is None:
        barras = ax1.bar(x, unified_table[emo], label=emo, color=col)
        bottom = unified_table[emo].copy()
    else:
        barras = ax1.bar(x, unified_table[emo], bottom=bottom, label=emo, color=col)
        bottom += unified_table[emo]

ax1.set_ylabel('Número de canciones por emoción')
ax1.set_xlabel('Año')
ax1.set_xticks(x)
ax1.set_xticklabels(x, rotation=90)
ax1.grid(axis='y', linestyle='--', alpha=0.5)

ax2 = ax1.twinx()
linea, = ax2.plot(x, unified_table['Total_Victimas'], color='blue', marker='o', linestyle='-',
                  label='Total de víctimas')
ax2.set_ylabel('Total de víctimas')

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
fig.legend(handles1 + handles2, labels1 + labels2,
           loc="upper center", bbox_to_anchor=(0.5, -0.08),
           ncol=4, frameon=False)

plt.title('Total de víctimas y emociones de las producciones musicales por año (1990–2023)')
plt.tight_layout(rect=[0, 0.08, 1, 1])
plt.show()
```
